```python
import torch
```

```python
def sigmoid_activation(x):
  return 1 / (1 + torch.exp(-x))
```

```python
# create a network with:
# input layer with 5 neuron
# one hidden layer with one neuron
# output layer
torch.manual_seed(0)
features = torch.randn((1, 5))
weights = torch.rand_like(features)
bias = torch.randn((1, 1))
```

```python
# compute the output:
output = sigmoid_activation(torch.mm(features, torch.transpose(weights, 0, 1)) + bias)
print(output)
```
