```python
import torch
```

```python
def sigmoid_activation(x):
  return 1 / (1 + torch.exp(-x))
```

```python
# create a network with:
# input layer with 5 neuron
# one hidden layer with one neuron
# output layer
torch.manual_seed(0)
features = torch.randn((1, 5))
weights = torch.rand_like(features)
bias = torch.randn((1, 1))
```

```python
# compute the output:
output = sigmoid_activation(torch.mm(features, torch.transpose(weights, 0, 1)) + bias)
print(output)
```

```python
# create a network with:
# input layer with n_features neurons
# one hidden layer with n_neurons_hidden_layer
# output layer with one output
n_features = 20
n_neurons_hidden_layer = 5
n_output = 1

features = torch.randn((1, n_features))

weights_first_layer = torch.randn((n_features, n_neurons_hidden_layer)) # initialize weights from the input layer to the hidden layer
weights_second_layer = torch.randn((n_neurons_hidden_layer, n_output)) # initialize weights from the hidden layer to the output layer

biases_first_layer = torch.randn((1, n_neurons_hidden_layer))
biases_second_layer = torch.randn((1, n_output))

output_first_layer = sigmoid_activation(torch.mm(features, weights_first_layer) + biases_first_layer)
output = sigmoid_activation(torch.mm(output_first_layer, weights_second_layer) + biases_second_layer)

print(output)
```
